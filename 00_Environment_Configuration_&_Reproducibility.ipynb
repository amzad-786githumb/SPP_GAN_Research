```python
# ==============================================================================
# 00.4 — REQUIRED & OPTIONAL PACKAGE VERIFICATION
# ==============================================================================
#
# PURPOSE
# -------
# Verify package availability, importability, and installed versions for the
# research environment.
#
# REPRODUCIBILITY
# ---------------
# Package versions are captured in PACKAGE_STATUS and PACKAGE_VERSIONS so
# that Section 16/17/19 can persist the computational environment used by
# the experiment.
#
# POLICY
# ------
# - Required packages must be installed and importable.
# - Optional packages are recorded but do not stop Notebook 00.
# - Version information is captured from the installed distribution.
# - Package discovery alone is not considered sufficient verification.
# - No package is installed or upgraded by this section.
#
# ==============================================================================

import importlib
import importlib.util
from importlib.metadata import version as package_version
from importlib.metadata import PackageNotFoundError


print("=" * 100)
print("00.4 — PACKAGE VERIFICATION")
print("=" * 100)


# ==============================================================================
# REQUIRED PACKAGES
# ==============================================================================

REQUIRED_PACKAGES = {

    "numpy": "numpy",

    "pandas": "pandas",

    "scipy": "scipy",

    "sklearn": "scikit-learn",

    "matplotlib": "matplotlib",

    "seaborn": "seaborn",

    "joblib": "joblib",

}


# ==============================================================================
# OPTIONAL / DOWNSTREAM PACKAGES
# ==============================================================================

OPTIONAL_PACKAGES = {

    "torch": "torch",

    "sdmetrics": "sdmetrics",

    "ctgan": "ctgan",

    "opacus": "opacus",

}


# ==============================================================================
# PACKAGE STATUS CONTAINERS
# ==============================================================================

PACKAGE_STATUS = {}

PACKAGE_VERSIONS = {}

PACKAGE_IMPORT_ERRORS = {}


# ==============================================================================
# PACKAGE VERIFICATION FUNCTION
# ==============================================================================

def verify_package(
    import_name,
    distribution_name,
    required
):
    """
    Verify package discovery, importability, and installed version.
    """

    status = {

        "import_name":
            import_name,

        "distribution_name":
            distribution_name,

        "required":
            bool(required),

        "available":
            False,

        "importable":
            False,

        "version":
            None,

        "error":
            None,

    }


    # --------------------------------------------------------------------------
    # 1. Package discovery
    # --------------------------------------------------------------------------

    try:

        available = (
            importlib.util.find_spec(
                import_name
            )
            is not None
        )

    except Exception as exc:

        available = False

        status["error"] = (
            f"Package discovery failed: {exc}"
        )


    status["available"] = bool(
        available
    )


    # --------------------------------------------------------------------------
    # 2. Installed distribution version
    # --------------------------------------------------------------------------

    if status["available"]:

        try:

            status["version"] = (
                package_version(
                    distribution_name
                )
            )

        except PackageNotFoundError:

            status["version"] = None

            status["error"] = (
                "Distribution metadata not found."
            )

        except Exception as exc:

            status["version"] = None

            status["error"] = (
                f"Version lookup failed: {exc}"
            )


    # --------------------------------------------------------------------------
    # 3. Actual import verification
    # --------------------------------------------------------------------------

    if status["available"]:

        try:

            importlib.import_module(
                import_name
            )

            status["importable"] = True

        except Exception as exc:

            status["importable"] = False

            status["error"] = (
                f"Import failed: {type(exc).__name__}: {exc}"
            )


    return status


# ==============================================================================
# VERIFY REQUIRED PACKAGES
# ==============================================================================

print()
print("REQUIRED PACKAGES")
print("-" * 100)


for import_name, distribution_name in REQUIRED_PACKAGES.items():

    status = verify_package(
        import_name=import_name,
        distribution_name=distribution_name,
        required=True,
    )

    PACKAGE_STATUS[
        distribution_name
    ] = status


    PACKAGE_VERSIONS[
        distribution_name
    ] = status["version"]


    if status["error"] is not None:

        PACKAGE_IMPORT_ERRORS[
            distribution_name
        ] = status["error"]


    if (
        status["available"]
        and
        status["importable"]
    ):

        print(
            f"{distribution_name:<20} : "
            f"AVAILABLE | "
            f"version={status['version']}"
        )

    elif status["available"]:

        print(
            f"{distribution_name:<20} : "
            f"INSTALLED BUT IMPORT FAILED"
        )

    else:

        print(
            f"{distribution_name:<20} : "
            f"MISSING"
        )


# ==============================================================================
# VERIFY OPTIONAL PACKAGES
# ==============================================================================

print()
print("OPTIONAL / DOWNSTREAM PACKAGES")
print("-" * 100)


for import_name, distribution_name in OPTIONAL_PACKAGES.items():

    status = verify_package(
        import_name=import_name,
        distribution_name=distribution_name,
        required=False,
    )

    PACKAGE_STATUS[
        distribution_name
    ] = status


    PACKAGE_VERSIONS[
        distribution_name
    ] = status["version"]


    if status["error"] is not None:

        PACKAGE_IMPORT_ERRORS[
            distribution_name
        ] = status["error"]


    if (
        status["available"]
        and
        status["importable"]
    ):

        print(
            f"{distribution_name:<20} : "
            f"AVAILABLE | "
            f"version={status['version']}"
        )

    elif status["available"]:

        print(
            f"{distribution_name:<20} : "
            f"INSTALLED BUT IMPORT FAILED"
        )

    else:

        print(
            f"{distribution_name:<20} : "
            f"NOT AVAILABLE"
        )


# ==============================================================================
# REQUIRED PACKAGE INTEGRITY CHECK
# ==============================================================================

REQUIRED_PACKAGE_CHECKS = {}


for distribution_name in REQUIRED_PACKAGES.values():

    status = PACKAGE_STATUS[
        distribution_name
    ]


    REQUIRED_PACKAGE_CHECKS[
        f"{distribution_name}_available"
    ] = (
        status["available"]
    )


    REQUIRED_PACKAGE_CHECKS[
        f"{distribution_name}_importable"
    ] = (
        status["importable"]
    )


    REQUIRED_PACKAGE_CHECKS[
        f"{distribution_name}_version_available"
    ] = (
        status["version"] is not None
    )


# ==============================================================================
# PRINT REQUIRED PACKAGE CHECKS
# ==============================================================================

print()
print("REQUIRED PACKAGE INTEGRITY")
print("-" * 100)


for check_name, passed in REQUIRED_PACKAGE_CHECKS.items():

    print(
        f"  {'PASS' if passed else 'FAIL':<6} "
        f"{check_name}"
    )


PACKAGE_VERIFICATION_PASS = all(
    REQUIRED_PACKAGE_CHECKS.values()
)


# ==============================================================================
# FAILURE GATE
# ==============================================================================

if not PACKAGE_VERIFICATION_PASS:

    failed_checks = [
        check_name
        for check_name, passed
        in REQUIRED_PACKAGE_CHECKS.items()
        if not passed
    ]


    error_details = []


    for distribution_name, error in PACKAGE_IMPORT_ERRORS.items():

        error_details.append(
            f"{distribution_name}: {error}"
        )


    message = (
        "Required package verification failed.\n\n"
        "Failed checks:\n"
        +
        "\n".join(
            f"  - {check}"
            for check in failed_checks
        )
    )


    if error_details:

        message += (
            "\n\nPackage errors:\n"
            +
            "\n".join(
                f"  - {error}"
                for error in error_details
            )
        )


    raise ImportError(
        message
    )


# ==============================================================================
# PACKAGE SUMMARY
# ==============================================================================

AVAILABLE_OPTIONAL_PACKAGES = [
    distribution_name
    for distribution_name in OPTIONAL_PACKAGES.values()
    if PACKAGE_STATUS[
        distribution_name
    ]["available"]
    and PACKAGE_STATUS[
        distribution_name
    ]["importable"]
]


MISSING_OPTIONAL_PACKAGES = [
    distribution_name
    for distribution_name in OPTIONAL_PACKAGES.values()
    if not (
        PACKAGE_STATUS[
            distribution_name
        ]["available"]
        and
        PACKAGE_STATUS[
            distribution_name
        ]["importable"]
    )
]


# ==============================================================================
# FINAL OUTPUT
# ==============================================================================

print()
print("=" * 100)
print("PACKAGE VERIFICATION SUMMARY")
print("=" * 100)

print(
    f"Required packages verified : "
    f"{len(REQUIRED_PACKAGES)}"
)

print(
    f"Required package status    : PASS"
)

print(
    f"Optional packages available: "
    f"{len(AVAILABLE_OPTIONAL_PACKAGES)}"
)

print(
    f"Optional packages absent   : "
    f"{len(MISSING_OPTIONAL_PACKAGES)}"
)

print()
print("✓ REQUIRED PACKAGE VERIFICATION STATUS : PASS")
print("=" * 100)
```

```python
# ==============================================================================
# 00.6 — GLOBAL RANDOM SEED CONFIGURATION
# ==============================================================================
#
# PURPOSE
# -------
# Establish the authoritative deterministic seed hierarchy for the complete
# SPP-GAN research pipeline.
#
# POLICY
# ------
# - MASTER_SEED is the single experiment-level seed.
# - Derived seeds are deterministic offsets from MASTER_SEED.
# - Downstream notebooks must consume these predefined seeds.
# - No downstream notebook should silently create a new master seed.
# - Python, NumPy and PyTorch RNGs are initialized here when available.
# - CUDA RNGs are initialized when CUDA is available.
#
# ==============================================================================

import random
import numpy as np


# ==============================================================================
# 1 — MASTER REPRODUCIBILITY SEED
# ==============================================================================

MASTER_SEED = 2025


if not isinstance(
    MASTER_SEED,
    int
):
    raise TypeError(
        "MASTER_SEED must be an integer."
    )


if MASTER_SEED < 0:
    raise ValueError(
        "MASTER_SEED must be non-negative."
    )


# ==============================================================================
# 2 — DETERMINISTIC SEED OFFSETS
# ==============================================================================

SEED_OFFSETS = {

    "numpy":
        0,

    "python":
        1,

    "torch":
        2,

    "data_split":
        10,

    "baseline":
        20,

    "sppgan":
        30,

    "generation":
        40,

    "evaluation":
        50,

    "ablation":
        60,

}


# ==============================================================================
# 3 — DERIVED SEEDS
# ==============================================================================

NUMPY_SEED = (
    MASTER_SEED
    + SEED_OFFSETS["numpy"]
)

PYTHON_SEED = (
    MASTER_SEED
    + SEED_OFFSETS["python"]
)

TORCH_SEED = (
    MASTER_SEED
    + SEED_OFFSETS["torch"]
)

DATA_SPLIT_SEED = (
    MASTER_SEED
    + SEED_OFFSETS["data_split"]
)

BASELINE_SEED = (
    MASTER_SEED
    + SEED_OFFSETS["baseline"]
)

SPPGAN_SEED = (
    MASTER_SEED
    + SEED_OFFSETS["sppgan"]
)

GENERATION_SEED = (
    MASTER_SEED
    + SEED_OFFSETS["generation"]
)

EVALUATION_SEED = (
    MASTER_SEED
    + SEED_OFFSETS["evaluation"]
)

ABLATION_SEED = (
    MASTER_SEED
    + SEED_OFFSETS["ablation"]
)


# ==============================================================================
# 4 — VALIDATE SEED HIERARCHY
# ==============================================================================

SEED_REGISTRY = {

    "master":
        MASTER_SEED,

    "python":
        PYTHON_SEED,

    "numpy":
        NUMPY_SEED,

    "torch":
        TORCH_SEED,

    "data_split":
        DATA_SPLIT_SEED,

    "baseline":
        BASELINE_SEED,

    "sppgan":
        SPPGAN_SEED,

    "generation":
        GENERATION_SEED,

    "evaluation":
        EVALUATION_SEED,

    "ablation":
        ABLATION_SEED,

}


SEED_CHECKS = {

    "master_seed_integer":
        isinstance(
            MASTER_SEED,
            int
        ),

    "master_seed_non_negative":
        MASTER_SEED >= 0,

    "numpy_seed_correct":
        NUMPY_SEED
        == MASTER_SEED
        + SEED_OFFSETS["numpy"],

    "python_seed_correct":
        PYTHON_SEED
        == MASTER_SEED
        + SEED_OFFSETS["python"],

    "torch_seed_correct":
        TORCH_SEED
        == MASTER_SEED
        + SEED_OFFSETS["torch"],

    "data_split_seed_correct":
        DATA_SPLIT_SEED
        == MASTER_SEED
        + SEED_OFFSETS["data_split"],

    "baseline_seed_correct":
        BASELINE_SEED
        == MASTER_SEED
        + SEED_OFFSETS["baseline"],

    "sppgan_seed_correct":
        SPPGAN_SEED
        == MASTER_SEED
        + SEED_OFFSETS["sppgan"],

    "generation_seed_correct":
        GENERATION_SEED
        == MASTER_SEED
        + SEED_OFFSETS["generation"],

    "evaluation_seed_correct":
        EVALUATION_SEED
        == MASTER_SEED
        + SEED_OFFSETS["evaluation"],

    "ablation_seed_correct":
        ABLATION_SEED
        == MASTER_SEED
        + SEED_OFFSETS["ablation"],

}


if not all(
    SEED_CHECKS.values()
):

    failed_checks = [
        check_name
        for check_name, passed
        in SEED_CHECKS.items()
        if not passed
    ]

    raise RuntimeError(
        "Seed hierarchy validation failed:\n"
        + "\n".join(
            f"  - {check_name}"
            for check_name in failed_checks
        )
    )


# ==============================================================================
# 5 — INITIALIZE PYTHON RNG
# ==============================================================================

random.seed(
    PYTHON_SEED
)


# ==============================================================================
# 6 — INITIALIZE NUMPY RNG
# ==============================================================================

np.random.seed(
    NUMPY_SEED
)


# ==============================================================================
# 7 — INITIALIZE PYTORCH RNG WHEN AVAILABLE
# ==============================================================================

TORCH_AVAILABLE = False
TORCH_CUDA_AVAILABLE = False
TORCH_VERSION = None

try:

    import torch

    TORCH_AVAILABLE = True

    TORCH_VERSION = getattr(
        torch,
        "__version__",
        None
    )

    torch.manual_seed(
        TORCH_SEED
    )

    TORCH_CUDA_AVAILABLE = bool(
        torch.cuda.is_available()
    )

    if TORCH_CUDA_AVAILABLE:

        torch.cuda.manual_seed(
            TORCH_SEED
        )

        torch.cuda.manual_seed_all(
            TORCH_SEED
        )

except Exception as exc:

    TORCH_AVAILABLE = False

    TORCH_CUDA_AVAILABLE = False

    TORCH_VERSION = None

    TORCH_INITIALIZATION_ERROR = (
        f"{type(exc).__name__}: {exc}"
    )


# ==============================================================================
# 8 — REPORT SEED CONFIGURATION
# ==============================================================================

print("=" * 100)
print("00.6 — GLOBAL RANDOM SEED CONFIGURATION")
print("=" * 100)

print(
    f"MASTER_SEED       : {MASTER_SEED}"
)

print(
    f"Python seed       : {PYTHON_SEED}"
)

print(
    f"NumPy seed        : {NUMPY_SEED}"
)

print(
    f"PyTorch seed      : {TORCH_SEED}"
)

print(
    f"Data split seed   : {DATA_SPLIT_SEED}"
)

print(
    f"Baseline seed     : {BASELINE_SEED}"
)

print(
    f"SPP-GAN seed      : {SPPGAN_SEED}"
)

print(
    f"Generation seed   : {GENERATION_SEED}"
)

print(
    f"Evaluation seed   : {EVALUATION_SEED}"
)

print(
    f"Ablation seed     : {ABLATION_SEED}"
)

print()
print(
    f"PyTorch available : {TORCH_AVAILABLE}"
)

print(
    f"CUDA available    : {TORCH_CUDA_AVAILABLE}"
)

if TORCH_VERSION is not None:

    print(
        f"PyTorch version   : {TORCH_VERSION}"
    )


# ==============================================================================
# 9 — FINAL STATUS
# ==============================================================================

print()
print(
    "✓ Python random seed initialized."
)

print(
    "✓ NumPy random seed initialized."
)

if TORCH_AVAILABLE:

    print(
        "✓ PyTorch random seed initialized."
    )

    if TORCH_CUDA_AVAILABLE:

        print(
            "✓ CUDA random seeds initialized."
        )

else:

    print(
        "⚠ PyTorch seed initialization skipped "
        "(PyTorch unavailable)."
    )


print()
print(
    "✓ GLOBAL RANDOM SEED CONFIGURATION STATUS : PASS"
)

print("=" * 100)
```

```python
# ==============================================================================
# 00.7 — REPRODUCIBILITY ENVIRONMENT CONFIGURATION
# ==============================================================================
#
# PURPOSE
# -------
# Establish the authoritative runtime-level reproducibility configuration for
# the complete SPP-GAN research pipeline.
#
# POLICY
# ------
# - Use the seed hierarchy established in Section 00.6.
# - Use TORCH_SEED for PyTorch and CUDA RNGs.
# - Control CPU thread usage for predictable Colab resource consumption.
# - Request deterministic PyTorch execution.
# - Disable cuDNN benchmarking to avoid algorithm-selection variability.
# - Record all reproducibility settings for later persistence.
#
# IMPORTANT
# ---------
# PYTHONHASHSEED is recorded and exported here, but changing the environment
# variable inside an already-running Python process does not retroactively
# change that process's hash randomization.
#
# ==============================================================================

import os
import importlib
from datetime import datetime, timezone


# ==============================================================================
# 1 — PYTHON HASH SEED
# ==============================================================================

PYTHONHASHSEED_VALUE = str(
    MASTER_SEED
)

os.environ[
    "PYTHONHASHSEED"
] = PYTHONHASHSEED_VALUE


# ==============================================================================
# 2 — CPU THREADING CONTROL
# ==============================================================================

OMP_NUM_THREADS_VALUE = "1"
MKL_NUM_THREADS_VALUE = "1"

os.environ[
    "OMP_NUM_THREADS"
] = OMP_NUM_THREADS_VALUE

os.environ[
    "MKL_NUM_THREADS"
] = MKL_NUM_THREADS_VALUE


# ==============================================================================
# 3 — PYTORCH AVAILABILITY
# ==============================================================================

TORCH_AVAILABLE = (
    importlib.util.find_spec("torch") is not None
)

TORCH_VERSION = None
TORCH_CUDA_AVAILABLE = False
TORCH_DETERMINISTIC_ALGORITHMS_ENABLED = False
TORCH_CUDNN_DETERMINISTIC = None
TORCH_CUDNN_BENCHMARK = None
TORCH_REPRODUCIBILITY_ERROR = None


# ==============================================================================
# 4 — PYTORCH REPRODUCIBILITY CONFIGURATION
# ==============================================================================

TORCH_REPRODUCIBILITY_CONFIG = {
    "enabled":
        TORCH_AVAILABLE,

    "seed":
        TORCH_SEED,

    "deterministic":
        True,

    "benchmark":
        False,

    "cuda_seeded":
        False,

    "cuda_deterministic":
        False,

    "deterministic_algorithms":
        False,
}


if TORCH_AVAILABLE:

    try:

        import torch

        TORCH_VERSION = getattr(
            torch,
            "__version__",
            None
        )

        # ----------------------------------------------------------------------
        # PyTorch CPU RNG
        # ----------------------------------------------------------------------

        torch.manual_seed(
            TORCH_SEED
        )

        # ----------------------------------------------------------------------
        # CUDA availability
        # ----------------------------------------------------------------------

        TORCH_CUDA_AVAILABLE = bool(
            torch.cuda.is_available()
        )

        # ----------------------------------------------------------------------
        # CUDA RNG
        # ----------------------------------------------------------------------

        if TORCH_CUDA_AVAILABLE:

            torch.cuda.manual_seed(
                TORCH_SEED
            )

            torch.cuda.manual_seed_all(
                TORCH_SEED
            )

            TORCH_REPRODUCIBILITY_CONFIG[
                "cuda_seeded"
            ] = True

        # ----------------------------------------------------------------------
        # cuDNN deterministic execution
        # ----------------------------------------------------------------------

        if hasattr(
            torch.backends,
            "cudnn"
        ):

            torch.backends.cudnn.deterministic = True

            torch.backends.cudnn.benchmark = False

            TORCH_CUDNN_DETERMINISTIC = bool(
                torch.backends.cudnn.deterministic
            )

            TORCH_CUDNN_BENCHMARK = bool(
                torch.backends.cudnn.benchmark
            )

            TORCH_REPRODUCIBILITY_CONFIG[
                "cuda_deterministic"
            ] = TORCH_CUDNN_DETERMINISTIC

        # ----------------------------------------------------------------------
        # Strict deterministic algorithms
        # ----------------------------------------------------------------------

        try:

            torch.use_deterministic_algorithms(
                True
            )

            TORCH_DETERMINISTIC_ALGORITHMS_ENABLED = True

            TORCH_REPRODUCIBILITY_CONFIG[
                "deterministic_algorithms"
            ] = True

        except Exception as exc:

            TORCH_REPRODUCIBILITY_ERROR = (
                f"{type(exc).__name__}: {exc}"
            )

    except Exception as exc:

        TORCH_REPRODUCIBILITY_ERROR = (
            f"{type(exc).__name__}: {exc}"
        )


# ==============================================================================
# 5 — BUILD AUTHORITATIVE REPRODUCIBILITY CONFIGURATION
# ==============================================================================

REPRODUCIBILITY_CONFIG = {

    "master_seed":
        MASTER_SEED,

    "python_seed":
        PYTHON_SEED,

    "numpy_seed":
        NUMPY_SEED,

    "torch_seed":
        TORCH_SEED,

    "data_split_seed":
        DATA_SPLIT_SEED,

    "baseline_seed":
        BASELINE_SEED,

    "sppgan_seed":
        SPPGAN_SEED,

    "generation_seed":
        GENERATION_SEED,

    "evaluation_seed":
        EVALUATION_SEED,

    "ablation_seed":
        ABLATION_SEED,

    "python_hash_seed":
        MASTER_SEED,

    "python_hash_seed_note":
        (
            "PYTHONHASHSEED is exported to the environment. "
            "Changing it after Python startup does not retroactively "
            "change hash randomization in the current process."
        ),

    "omp_num_threads":
        int(OMP_NUM_THREADS_VALUE),

    "mkl_num_threads":
        int(MKL_NUM_THREADS_VALUE),

    "torch_available":
        TORCH_AVAILABLE,

    "torch_version":
        TORCH_VERSION,

    "torch_cuda_available":
        TORCH_CUDA_AVAILABLE,

    "torch_deterministic_algorithms":
        TORCH_DETERMINISTIC_ALGORITHMS_ENABLED,

    "torch_cudnn_deterministic":
        TORCH_CUDNN_DETERMINISTIC,

    "torch_cudnn_benchmark":
        TORCH_CUDNN_BENCHMARK,

    "torch_reproducibility_error":
        TORCH_REPRODUCIBILITY_ERROR,

    "configured_at_utc":
        datetime.now(
            timezone.utc
        ).isoformat(),

}


# ==============================================================================
# 6 — VALIDATION CHECKS
# ==============================================================================

REPRODUCIBILITY_CHECKS = {

    "python_hash_seed_configured":
        os.environ.get(
            "PYTHONHASHSEED"
        ) == str(MASTER_SEED),

    "omp_threads_configured":
        os.environ.get(
            "OMP_NUM_THREADS"
        ) == "1",

    "mkl_threads_configured":
        os.environ.get(
            "MKL_NUM_THREADS"
        ) == "1",

    "torch_seed_consistent":
        (
            not TORCH_AVAILABLE
            or TORCH_REPRODUCIBILITY_CONFIG[
                "seed"
            ] == TORCH_SEED
        ),

    "torch_deterministic":
        (
            not TORCH_AVAILABLE
            or TORCH_DETERMINISTIC_ALGORITHMS_ENABLED
        ),

    "cudnn_deterministic":
        (
            not TORCH_CUDA_AVAILABLE
            or TORCH_CUDNN_DETERMINISTIC is True
        ),

    "cudnn_benchmark_disabled":
        (
            not TORCH_CUDA_AVAILABLE
            or TORCH_CUDNN_BENCHMARK is False
        ),

}


# ==============================================================================
# 7 — FINAL VALIDATION
# ==============================================================================

FAILED_REPRODUCIBILITY_CHECKS = [

    check_name

    for check_name, passed
    in REPRODUCIBILITY_CHECKS.items()

    if not passed

]


REPRODUCIBILITY_CONFIGURATION_PASS = (
    len(
        FAILED_REPRODUCIBILITY_CHECKS
    ) == 0
)


if not REPRODUCIBILITY_CONFIGURATION_PASS:

    raise RuntimeError(
        "Reproducibility configuration validation failed:\n"
        + "\n".join(
            f"  - {check_name}"
            for check_name
            in FAILED_REPRODUCIBILITY_CHECKS
        )
    )


# ==============================================================================
# 8 — REPORT
# ==============================================================================

print("=" * 100)
print("00.7 — REPRODUCIBILITY CONFIGURATION")
print("=" * 100)

print(
    f"PYTHONHASHSEED       : "
    f"{os.environ['PYTHONHASHSEED']}"
)

print(
    f"OMP_NUM_THREADS      : "
    f"{os.environ['OMP_NUM_THREADS']}"
)

print(
    f"MKL_NUM_THREADS      : "
    f"{os.environ['MKL_NUM_THREADS']}"
)

print(
    f"Python seed          : "
    f"{PYTHON_SEED}"
)

print(
    f"NumPy seed           : "
    f"{NUMPY_SEED}"
)

print(
    f"PyTorch seed         : "
    f"{TORCH_SEED}"
)

print(
    f"PyTorch available    : "
    f"{TORCH_AVAILABLE}"
)

if TORCH_AVAILABLE:

    print(
        f"PyTorch version      : "
        f"{TORCH_VERSION}"
    )

    print(
        f"CUDA available       : "
        f"{TORCH_CUDA_AVAILABLE}"
    )

    print(
        f"Deterministic alg.   : "
        f"{TORCH_DETERMINISTIC_ALGORITHMS_ENABLED}"
    )

    print(
        f"cuDNN deterministic   : "
        f"{TORCH_CUDNN_DETERMINISTIC}"
    )

    print(
        f"cuDNN benchmark       : "
        f"{TORCH_CUDNN_BENCHMARK}"
    )


print()
print(
    "✓ Reproducibility configuration initialized."
)

print(
    "✓ Seed hierarchy aligned with Section 00.6."
)

print(
    "✓ CPU threading constrained."
)

if TORCH_AVAILABLE:

    print(
        "✓ PyTorch reproducibility controls configured."
    )

if TORCH_CUDA_AVAILABLE:

    print(
        "✓ CUDA reproducibility controls configured."
    )

print(
    "✓ REPRODUCIBILITY CONFIGURATION STATUS : PASS"
)

print("=" * 100)
```

```python
# ==============================================================================
# 00.8 — EXPERIMENTAL CONFIGURATION
# ==============================================================================
#
# PURPOSE
# -------
# Define the authoritative experimental design for the SPP-GAN research study.
#
# POLICY
# ------
# - Three canonical datasets are evaluated.
# - Five deterministic experimental repetitions are used.
# - Train/validation/test fractions are fixed at 70/15/15.
# - Generative models and preprocessing are fitted using TRAIN data only.
# - Validation and TEST data remain reserved for downstream evaluation.
# - Comparison methods use stable machine-readable identifiers.
# - Every repetition receives a deterministic derived seed.
# - Downstream notebooks must consume this configuration rather than
#   independently redefining the experimental design.
#
# ==============================================================================


# ==============================================================================
# 1 — CANONICAL DATASET REGISTRY FOR EXPERIMENTS
# ==============================================================================

DATASET_IDS = [
    "adult_income",
    "bank_marketing",
    "diabetes_130us",
]


# ==============================================================================
# 2 — EXPERIMENTAL REPETITIONS
# ==============================================================================

REPETITIONS = 5


if not isinstance(
    REPETITIONS,
    int
):

    raise TypeError(
        "REPETITIONS must be an integer."
    )


if REPETITIONS < 1:

    raise ValueError(
        "REPETITIONS must be at least 1."
    )


# ==============================================================================
# 3 — DETERMINISTIC REPETITION SEED POLICY
# ==============================================================================

REPETITION_SEED_OFFSET = 1000

REPETITION_SEEDS = {
    repetition:
        MASTER_SEED
        + REPETITION_SEED_OFFSET
        + repetition

    for repetition
    in range(
        1,
        REPETITIONS + 1
    )
}


# Example:
# repetition 1 -> 3026
# repetition 2 -> 3027
# ...
# repetition 5 -> 3030


SEED_POLICY = {

    "master_seed":
        MASTER_SEED,

    "repetitions":
        REPETITIONS,

    "deterministic":
        True,

    "repetition_seed_offset":
        REPETITION_SEED_OFFSET,

    "repetition_seeds":
        REPETITION_SEEDS,

}


# ==============================================================================
# 4 — DATA SPLIT CONFIGURATION
# ==============================================================================

DATA_SPLIT_CONFIG = {

    "enabled":
        True,

    "train_fraction":
        0.70,

    "validation_fraction":
        0.15,

    "test_fraction":
        0.15,

    "random_state":
        DATA_SPLIT_SEED,

    "fit_split":
        "train",

    "validation_role":
        "downstream_evaluation",

    "test_role":
        "final_evaluation",

}


# ==============================================================================
# 5 — VALIDATE DATA SPLIT CONFIGURATION
# ==============================================================================

SPLIT_FRACTIONS = [

    DATA_SPLIT_CONFIG[
        "train_fraction"
    ],

    DATA_SPLIT_CONFIG[
        "validation_fraction"
    ],

    DATA_SPLIT_CONFIG[
        "test_fraction"
    ],

]


if not all(
    isinstance(
        value,
        (int, float)
    )
    and 0.0 < value < 1.0
    for value in SPLIT_FRACTIONS
):

    raise ValueError(
        "All data split fractions must be between 0 and 1."
    )


if abs(
    sum(SPLIT_FRACTIONS)
    - 1.0
) >= 1e-12:

    raise ValueError(
        "Train, validation and test fractions must sum to 1.0."
    )


# ==============================================================================
# 6 — CANONICAL COMPARISON MODEL REGISTRY
# ==============================================================================

COMPARISON_MODEL_REGISTRY = {

    "statistical": {
        "display_name":
            "Statistical Baseline",
        "category":
            "statistical",
    },

    "tvae": {
        "display_name":
            "TVAE",
        "category":
            "machine_learning",
    },

    "ctgan": {
        "display_name":
            "CTGAN",
        "category":
            "machine_learning",
    },

    "dp_ctgan": {
        "display_name":
            "DP-CTGAN",
        "category":
            "privacy_preserving_machine_learning",
    },

    "spp_gan": {
        "display_name":
            "SPP-GAN",
        "category":
            "proposed_method",
    },

}


COMPARISON_MODEL_IDS = list(
    COMPARISON_MODEL_REGISTRY.keys()
)


# ==============================================================================
# 7 — EVALUATION REGISTRY
# ==============================================================================

EVALUATION_REGISTRY = {

    "statistical_fidelity":
        {
            "display_name":
                "Statistical Fidelity",
        },

    "ml_utility":
        {
            "display_name":
                "ML Utility",
        },

    "privacy_disclosure_risk":
        {
            "display_name":
                "Privacy & Disclosure Risk",
        },

}


ADVANCED_ANALYSIS_REGISTRY = {

    "ablation":
        {
            "display_name":
                "Ablation Study",
        },

    "privacy_fidelity_utility_tradeoff":
        {
            "display_name":
                "Privacy–Fidelity–Utility Trade-off",
        },

    "statistical_significance":
        {
            "display_name":
                "Statistical Significance Testing",
        },

}


# ==============================================================================
# 8 — EXPERIMENTAL UNIT
# ==============================================================================

EXPERIMENTAL_UNIT = {

    "primary_unit":
        "dataset × repetition × method",

    "evaluation_data_policy":
        {
            "training":
                "train",

            "validation":
                "validation",

            "final_test":
                "test",
        },

}


# ==============================================================================
# 9 — COMPLETE EXPERIMENT CONFIGURATION
# ==============================================================================

EXPERIMENT_CONFIG = {

    "datasets":
        DATASET_IDS,

    "repetitions":
        REPETITIONS,

    "seed_policy":
        SEED_POLICY,

    "data_split":
        DATA_SPLIT_CONFIG,

    "comparison_models":
        COMPARISON_MODEL_IDS,

    "comparison_model_registry":
        COMPARISON_MODEL_REGISTRY,

    "evaluation":
        list(
            EVALUATION_REGISTRY.keys()
        ),

    "evaluation_registry":
        EVALUATION_REGISTRY,

    "advanced_analysis":
        list(
            ADVANCED_ANALYSIS_REGISTRY.keys()
        ),

    "advanced_analysis_registry":
        ADVANCED_ANALYSIS_REGISTRY,

    "experimental_unit":
        EXPERIMENTAL_UNIT,

}


# ==============================================================================
# 10 — EXPERIMENT CONFIGURATION VALIDATION
# ==============================================================================

EXPERIMENT_CONFIG_CHECKS = {

    "dataset_count":
        len(DATASET_IDS) == 3,

    "all_datasets_unique":
        len(DATASET_IDS)
        == len(set(DATASET_IDS)),

    "repetitions_valid":
        REPETITIONS >= 1,

    "repetition_seed_count":
        len(REPETITION_SEEDS)
        == REPETITIONS,

    "repetition_seeds_unique":
        len(
            set(
                REPETITION_SEEDS.values()
            )
        )
        == REPETITIONS,

    "train_fraction":
        DATA_SPLIT_CONFIG[
            "train_fraction"
        ] == 0.70,

    "validation_fraction":
        DATA_SPLIT_CONFIG[
            "validation_fraction"
        ] == 0.15,

    "test_fraction":
        DATA_SPLIT_CONFIG[
            "test_fraction"
        ] == 0.15,

    "split_sum":
        abs(
            sum(SPLIT_FRACTIONS)
            - 1.0
        ) < 1e-12,

    "comparison_model_count":
        len(COMPARISON_MODEL_REGISTRY)
        == 5,

    "evaluation_count":
        len(EVALUATION_REGISTRY)
        == 3,

    "advanced_analysis_count":
        len(ADVANCED_ANALYSIS_REGISTRY)
        == 3,

}


FAILED_EXPERIMENT_CONFIG_CHECKS = [

    check_name

    for check_name, passed
    in EXPERIMENT_CONFIG_CHECKS.items()

    if not passed

]


EXPERIMENT_CONFIG_PASS = (
    len(
        FAILED_EXPERIMENT_CONFIG_CHECKS
    ) == 0
)


if not EXPERIMENT_CONFIG_PASS:

    raise RuntimeError(
        "Experimental configuration validation failed:\n"
        + "\n".join(
            f"  - {check_name}"
            for check_name
            in FAILED_EXPERIMENT_CONFIG_CHECKS
        )
    )


# ==============================================================================
# 11 — REPORT
# ==============================================================================

print("=" * 100)
print("00.8 — EXPERIMENTAL CONFIGURATION")
print("=" * 100)

print(
    f"Datasets        : {DATASET_IDS}"
)

print(
    f"Repetitions     : {REPETITIONS}"
)

print(
    f"Master seed     : {MASTER_SEED}"
)

print(
    f"Repetition seeds: {REPETITION_SEEDS}"
)

print(
    f"Data split      : "
    f"{DATA_SPLIT_CONFIG['train_fraction']:.2f} / "
    f"{DATA_SPLIT_CONFIG['validation_fraction']:.2f} / "
    f"{DATA_SPLIT_CONFIG['test_fraction']:.2f}"
)

print(
    f"Comparison      : {COMPARISON_MODEL_IDS}"
)

print(
    f"Evaluation      : "
    f"{list(EVALUATION_REGISTRY.keys())}"
)

print(
    f"Advanced        : "
    f"{list(ADVANCED_ANALYSIS_REGISTRY.keys())}"
)

print(
    f"Experimental unit : "
    f"{EXPERIMENTAL_UNIT['primary_unit']}"
)

print()
print(
    "✓ Experimental configuration validated."
)

print(
    "✓ Train-only fitting policy registered."
)

print(
    "✓ Deterministic repetition seed policy registered."
)

print(
    "✓ EXPERIMENTAL CONFIGURATION STATUS : PASS"
)

print("=" * 100)
```

```python
# ==============================================================================
# 00.18 — SAVE CONFIGURATION ARTIFACTS
# ==============================================================================

import json
from pathlib import Path

# ------------------------------------------------------------------------------
# 1. Deterministic JSON writer
# ------------------------------------------------------------------------------

def save_json(data, path):
    """
    Save JSON using the canonical Notebook 00 serialization policy.
    """

    path = Path(path)
    path.parent.mkdir(parents=True, exist_ok=True)

    with open(path, "w", encoding="utf-8") as f:
        json.dump(
            data,
            f,
            indent=2,
            sort_keys=True,
            default=str,
            ensure_ascii=False,
        )


# ------------------------------------------------------------------------------
# 2. Canonical JSON serialization helper
# ------------------------------------------------------------------------------

def canonical_json_string(data):
    """
    Produce the canonical JSON representation used for persistence
    validation.

    The same serialization policy is used by save_json().
    """

    return json.dumps(
        data,
        sort_keys=True,
        default=str,
        ensure_ascii=False,
        separators=(",", ":"),
    )


# ------------------------------------------------------------------------------
# 3. Authoritative artifact payload registry
# ------------------------------------------------------------------------------

ARTIFACT_PAYLOADS = {

    "config_json":
        EXPERIMENT_CONFIG,

    "dataset_registry_json":
        DATASET_REGISTRY,

    "model_registry_json":
        MODEL_REGISTRY,

    "evaluation_config_json":
        EVALUATION_CONFIG,

    "privacy_config_json":
        PRIVACY_CONFIG,

    "sppgan_config_json":
        SPPGAN_CONFIG,

    "environment_json":
        ENVIRONMENT_METADATA,

    "manifest_json":
        REPRODUCIBILITY_MANIFEST,
}


# ------------------------------------------------------------------------------
# 4. Validate artifact/path alignment
# ------------------------------------------------------------------------------

EXPECTED_ARTIFACT_KEYS = set(ARTIFACT_PATHS.keys())
PAYLOAD_ARTIFACT_KEYS = set(ARTIFACT_PAYLOADS.keys())

ARTIFACT_KEY_ALIGNMENT_PASS = (
    EXPECTED_ARTIFACT_KEYS == PAYLOAD_ARTIFACT_KEYS
)

if not ARTIFACT_KEY_ALIGNMENT_PASS:

    missing_payloads = sorted(
        EXPECTED_ARTIFACT_KEYS - PAYLOAD_ARTIFACT_KEYS
    )

    unexpected_payloads = sorted(
        PAYLOAD_ARTIFACT_KEYS - EXPECTED_ARTIFACT_KEYS
    )

    raise RuntimeError(
        "Artifact payload/path mismatch.\n"
        f"Missing payloads: {missing_payloads}\n"
        f"Unexpected payloads: {unexpected_payloads}"
    )


# ------------------------------------------------------------------------------
# 5. Persist all artifacts
# ------------------------------------------------------------------------------

for artifact_name, payload in ARTIFACT_PAYLOADS.items():

    save_json(
        payload,
        ARTIFACT_PATHS[artifact_name],
    )


# ------------------------------------------------------------------------------
# 6. Physical artifact verification
# ------------------------------------------------------------------------------

ARTIFACT_SAVE_STATUS = {}

for artifact_name, path in ARTIFACT_PATHS.items():

    path = Path(path)

    exists_pass = path.exists()
    file_pass = path.is_file() if exists_pass else False
    size_bytes = path.stat().st_size if exists_pass else 0
    non_empty_pass = size_bytes > 0

    ARTIFACT_SAVE_STATUS[artifact_name] = {
        "path": str(path),
        "exists": exists_pass,
        "is_file": file_pass,
        "size_bytes": size_bytes,
        "non_empty": non_empty_pass,
    }


# ------------------------------------------------------------------------------
# 7. JSON parse verification
# ------------------------------------------------------------------------------

ARTIFACT_VALIDATION = {}

for artifact_name, status in ARTIFACT_SAVE_STATUS.items():

    path = Path(status["path"])

    json_parse_pass = False

    if (
        status["exists"]
        and status["is_file"]
        and status["non_empty"]
    ):

        try:

            with open(path, "r", encoding="utf-8") as f:
                json.load(f)

            json_parse_pass = True

        except Exception:
            json_parse_pass = False

    ARTIFACT_VALIDATION[artifact_name] = {
        "exists": status["exists"],
        "is_file": status["is_file"],
        "non_empty": status["non_empty"],
        "json_parseable": json_parse_pass,
        "pass": (
            status["exists"]
            and status["is_file"]
            and status["non_empty"]
            and json_parse_pass
        ),
    }


# ------------------------------------------------------------------------------
# 8. Load persisted reproducibility manifest
# ------------------------------------------------------------------------------

MANIFEST_PATH = Path(
    ARTIFACT_PATHS["manifest_json"]
)

with open(MANIFEST_PATH, "r", encoding="utf-8") as f:
    PERSISTED_MANIFEST = json.load(f)


# ------------------------------------------------------------------------------
# 9. Validate manifest structure
# ------------------------------------------------------------------------------

MANIFEST_REQUIRED_KEYS = [
    "manifest_version",
    "project",
    "environment",
    "configuration",
    "artifact_paths",
]

MANIFEST_REQUIRED_KEYS_PASS = all(
    key in PERSISTED_MANIFEST
    for key in MANIFEST_REQUIRED_KEYS
)


# ------------------------------------------------------------------------------
# 10. Canonical configuration comparison
# ------------------------------------------------------------------------------

EXPECTED_CONFIGURATION_JSON = canonical_json_string(
    CONFIG_SNAPSHOT
)

PERSISTED_CONFIGURATION_JSON = canonical_json_string(
    PERSISTED_MANIFEST.get("configuration", {})
)

MANIFEST_CONFIGURATION_PASS = (
    PERSISTED_CONFIGURATION_JSON
    == EXPECTED_CONFIGURATION_JSON
)


# ------------------------------------------------------------------------------
# 11. Canonical environment comparison
# ------------------------------------------------------------------------------

EXPECTED_ENVIRONMENT_JSON = canonical_json_string(
    ENVIRONMENT_METADATA
)

PERSISTED_ENVIRONMENT_JSON = canonical_json_string(
    PERSISTED_MANIFEST.get("environment", {})
)

MANIFEST_ENVIRONMENT_PASS = (
    PERSISTED_ENVIRONMENT_JSON
    == EXPECTED_ENVIRONMENT_JSON
)


# ------------------------------------------------------------------------------
# 12. Artifact-path comparison
# ------------------------------------------------------------------------------

EXPECTED_ARTIFACT_PATHS = {
    key: str(value)
    for key, value in ARTIFACT_PATHS.items()
}

PERSISTED_ARTIFACT_PATHS = (
    PERSISTED_MANIFEST.get("artifact_paths", {})
)

MANIFEST_ARTIFACT_PATHS_PASS = (
    PERSISTED_ARTIFACT_PATHS
    == EXPECTED_ARTIFACT_PATHS
)


# ------------------------------------------------------------------------------
# 13. Validate manifest project linkage
# ------------------------------------------------------------------------------

MANIFEST_PROJECT_PASS = (
    PERSISTED_MANIFEST.get("project")
    == CONFIG_SNAPSHOT.get("project")
)


# ------------------------------------------------------------------------------
# 14. Final artifact validation
# ------------------------------------------------------------------------------

FAILED_ARTIFACTS = [
    artifact_name
    for artifact_name, validation in ARTIFACT_VALIDATION.items()
    if not validation["pass"]
]

ALL_ARTIFACTS_PERSISTED = (
    len(FAILED_ARTIFACTS) == 0
)


# ------------------------------------------------------------------------------
# 15. Final Section 18 status
# ------------------------------------------------------------------------------

SECTION_18_PASS = (

    ARTIFACT_KEY_ALIGNMENT_PASS

    and ALL_ARTIFACTS_PERSISTED

    and MANIFEST_REQUIRED_KEYS_PASS

    and MANIFEST_PROJECT_PASS

    and MANIFEST_CONFIGURATION_PASS

    and MANIFEST_ENVIRONMENT_PASS

    and MANIFEST_ARTIFACT_PATHS_PASS
)


# ------------------------------------------------------------------------------
# 16. Output
# ------------------------------------------------------------------------------

print("=" * 100)
print("00.18 — CONFIGURATION ARTIFACTS SAVED")
print("=" * 100)

print("\nArtifact persistence:")

for artifact_name, validation in ARTIFACT_VALIDATION.items():

    status = "PASS" if validation["pass"] else "FAIL"

    size_bytes = ARTIFACT_SAVE_STATUS[
        artifact_name
    ]["size_bytes"]

    print(
        f"{artifact_name:<30} : "
        f"{status} "
        f"({size_bytes} bytes)"
    )


print("\nManifest validation:")

print(
    f"  Required keys                 : "
    f"{'PASS' if MANIFEST_REQUIRED_KEYS_PASS else 'FAIL'}"
)

print(
    f"  Project linkage               : "
    f"{'PASS' if MANIFEST_PROJECT_PASS else 'FAIL'}"
)

print(
    f"  Configuration snapshot        : "
    f"{'PASS' if MANIFEST_CONFIGURATION_PASS else 'FAIL'}"
)

print(
    f"  Environment metadata          : "
    f"{'PASS' if MANIFEST_ENVIRONMENT_PASS else 'FAIL'}"
)

print(
    f"  Artifact paths                : "
    f"{'PASS' if MANIFEST_ARTIFACT_PATHS_PASS else 'FAIL'}"
)


print("\nPersistence summary:")

print(
    f"  Artifact/path alignment       : "
    f"{'PASS' if ARTIFACT_KEY_ALIGNMENT_PASS else 'FAIL'}"
)

print(
    f"  Total artifacts               : "
    f"{len(ARTIFACT_PATHS)}"
)

print(
    f"  Failed artifacts              : "
    f"{len(FAILED_ARTIFACTS)}"
)

print(
    f"  Overall status                : "
    f"{'PASS' if SECTION_18_PASS else 'FAIL'}"
)


# ------------------------------------------------------------------------------
# 17. Failure diagnostics
# ------------------------------------------------------------------------------

if FAILED_ARTIFACTS:

    print("\n✗ Failed artifacts:")

    for artifact_name in FAILED_ARTIFACTS:
        print(f"  - {artifact_name}")


if not MANIFEST_PROJECT_PASS:
    print("\n✗ Persisted manifest project metadata mismatch.")


if not MANIFEST_CONFIGURATION_PASS:
    print(
        "\n✗ Persisted manifest configuration does not match "
        "the canonical serialized CONFIG_SNAPSHOT."
    )


if not MANIFEST_ENVIRONMENT_PASS:
    print(
        "\n✗ Persisted manifest environment metadata does not "
        "match the canonical serialized ENVIRONMENT_METADATA."
    )


if not MANIFEST_ARTIFACT_PATHS_PASS:
    print(
        "\n✗ Persisted manifest artifact paths do not match "
        "ARTIFACT_PATHS."
    )


# ------------------------------------------------------------------------------
# 18. Final gate
# ------------------------------------------------------------------------------

if not SECTION_18_PASS:

    raise RuntimeError(
        "Notebook 00 configuration artifact persistence failed."
    )


print(
    "\n✓ All Notebook 00 configuration artifacts "
    "persisted and verified successfully."
)
```
